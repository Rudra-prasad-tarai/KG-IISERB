```python
from langchain_community.document_loaders import PlaywrightURLLoader

pages = [
    "https://dsestudents.iiserb.ac.in",
    "https://dsestudents.iiserb.ac.in/bs-ms-5th-year",
    "https://dsestudents.iiserb.ac.in/bs-ms-majors-3rd-year",
    "https://dsestudents.iiserb.ac.in/bs-ms-majors-4th-year"
]

loader = PlaywrightURLLoader(
    urls=pages,
    headless=True,
    remove_selectors=["nav", "footer"]  # optional cleanup
)

docs = loader.load()
```
